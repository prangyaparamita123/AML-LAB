```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 3, 5, 4], dtype=float)

print(f"Dataset:\nx = {x}\ny = {y}")
```

```python

N = len(x)
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x_squared = np.sum(x ** 2)

m = (N * sum_xy - sum_x * sum_y) / (N * sum_x_squared - (sum_x) ** 2)
c = (sum_y - m * sum_x) / N

print(f"Regression Line: y = {m:.3f}x + {c:.3f}")
```

```python
y_pred = m * x + c

mae = np.mean(np.abs(y - y_pred))

mse = np.mean((y - y_pred) ** 2)

rmse = np.sqrt(mse)

ss_tot = np.sum((y - np.mean(y)) ** 2)
ss_res = np.sum((y - y_pred) ** 2)
r_squared = 1 - (ss_res / ss_tot)

print(f"Evaluation Metrics:")
print(f"-------------------")
print(f"Mean Absolute Error (MAE)       : {mae:.4f}")
print(f"Mean Squared Error (MSE)        : {mse:.4f}")
print(f"Root Mean Squared Error (RMSE)  : {rmse:.4f}")
print(f"R-squared (R2)                 : {r_squared:.4f}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', label='Data Points', s=100, zorder=5)
plt.plot(x, y_pred, color='red', linewidth=2, label=f'Regression Line: y = {m:.2f}x + {c:.2f}')
plt.title('Linear Regression using Least Squares Method')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```
