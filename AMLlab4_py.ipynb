```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3], dtype=float)
y = np.array([2, 3, 4], dtype=float)

learning_rate = 0.1
iterations = 100

w = 0.0
b = 0.0

n = len(X)
cost_history = []

for i in range(iterations):

    y_pred = w * X + b

    cost = (1 / (2 * n)) * np.sum((y_pred - y) ** 2)
    cost_history.append(cost)

    dw = (1 / n) * np.sum((y_pred - y) * X)
    db = (1 / n) * np.sum(y_pred - y)

    w = w - learning_rate * dw
    b = b - learning_rate * db

print(f"Training completed!")
print(f"Final Weight (w): {w:.4f} (Expected: 1.0)")
print(f"Final Bias (b): {b:.4f} (Expected: 1.0)")
print(f"Final Cost: {cost_history[-1]:.6f}")
```

```python

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, y, color='red', marker='o', s=100, label='Training Data')
x_line = np.linspace(0, 4, 100)
y_line = w * x_line + b
plt.plot(x_line, y_line, color='blue', linewidth=2, label=f'Regression Line: y = {w:.2f}x + {b:.2f}')
plt.title('Training Data and Fitted Regression Line')
plt.xlabel('X (Feature)')
plt.ylabel('Y (Target)')
plt.xlim(0, 4)
plt.ylim(0, 6)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(range(1, iterations + 1), cost_history, color='green', linewidth=2)
plt.title('Cost Decreasing towards Global Minima')
plt.xlabel('Iterations')
plt.ylabel('Cost (MSE)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
